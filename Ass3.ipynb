```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv(
    'Sample - Superstore Ass3.csv',
    encoding='latin1'
)

print("First 5 rows:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nDescriptive Statistics:")
print(df.describe())
```

```python
mean_val = df['Sales'].mean()
median_val = df['Sales'].median()
mode_val = df['Sales'].mode()[0]

print("\n--- Central Tendency using Pandas ---")

print("Mean:", mean_val)
print("Median:", median_val)
print("Mode:", mode_val)
```

```python
sales_array = df['Sales'].to_numpy()

numpy_mean = np.mean(sales_array)
numpy_median = np.median(sales_array)
scipy_mode = stats.mode(sales_array, keepdims=True)

print("\n--- Central Tendency using NumPy / SciPy ---")

print("NumPy Mean:", numpy_mean)
print("NumPy Median:", numpy_median)
print("SciPy Mode:", scipy_mode.mode[0])
print("SciPy Mode Count:", scipy_mode.count[0])
```

```python
group_summary = df.groupby('Category')['Sales'].agg([
    'mean',
    'median',
    lambda x: x.mode()[0]
])

group_summary.columns = ['Mean', 'Median', 'Mode']

print("\n--- Central Tendency by Category ---")
print(group_summary)

```

```python
skewness = df['Sales'].skew()

print("\n--- Distribution Analysis ---")

print("Skewness:", skewness)

if mean_val > median_val:
    print("Distribution is right (positively) skewed")

elif mean_val < median_val:
    print("Distribution is left (negatively) skewed")

else:
    print("Distribution is approximately symmetric")
```
