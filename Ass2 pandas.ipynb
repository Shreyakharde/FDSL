```python
df_fill = df.copy()

df_fill["Marks"] = df_fill["Marks"].fillna(df_fill["Marks"].mean())

df_fill
```

```python
def performance(marks):
    if marks >= 85:
        return "Excellent"
    elif marks >= 70:
        return "Good"
    else:
        return "Needs Improvement"

df_fill["Performance"] = df_fill["Marks"].apply(performance)

df_fill
```

```python
df_fill["Result_Code"] = df_fill["Result"].map({
    "Pass": 1,
    "Fail": 0
})

df_fill
```

```python
df_fill["Age"] = df_fill["Age"].astype(float)

df_fill.dtypes
```

```python
df_fill = df_fill.rename(
    columns={"Marks": "Total_Marks"}
)

df_fill.head()
```

```python
df_fill.sort_values(
    by="Total_Marks",
    ascending=False
)
```

```python
# Create another DataFrame for merging/joining

student_extra = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "City": ["Pune", "Mumbai", "Nashik", "Pune",
             "Mumbai", "Pune", "Nashik", "Mumbai"],
    "Course": ["B.Tech", "B.Tech", "B.Tech", "B.Tech",
               "B.Tech", "B.Tech", "B.Tech", "B.Tech"]
})

student_extra
```

```python
df1 = df.set_index("Student_ID")
df2 = student_extra.set_index("Student_ID")

joined_df = df1.join(df2[["City", "Course"]])

joined_df
```
